```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import rcParams
rcParams['font.family']='SimHei'
df=pd.DataFrame({"姓名":['小明','小红','李华'],"第一次月考":[557,567,589],"第二次月考":[546,589,605],"第三次月考":[520,600,598]})
df=df.set_index("姓名")
df['平均分']=df[['第一次月考','第二次月考','第三次月考']].mean(axis=1).round(2)
df['最高分']=df[['第一次月考','第二次月考','第三次月考']].max(axis=1)
df['最低分']=df[['第一次月考','第二次月考','第三次月考']].min(axis=1)
plt.figure(figsize=(10,5))
plt.plot(df.iloc[0,0:3],label='小明',marker='o')
plt.plot(df.iloc[1,0:3],label='小红',marker='o')
plt.plot(df.iloc[2,0:3],label='李华',marker='o')
plt.title('成绩走向',color='red',fontsize=15)
plt.ylabel('成绩',fontsize=10)
plt.xlabel('考试',fontsize=10)
plt.legend(loc='upper left')
plt.grid(axis='y',alpha=0.5)
plt.savefig("scores.png",dpi=300,bbox_inches="tight")
plt.show()
```

```python
f1=plt.subplot(1,3,1)
plt.bar(df.index,df['第一次月考'],width=0.8)
plt.title('第一次月考成绩对比',color='red',fontsize=13)
plt.ylim(500,610)
plt.tight_layout()
f2=plt.subplot(1,3,2)
plt.bar(df.index,df['第二次月考'],width=0.8)
plt.title('第二次月考成绩对比',color='red',fontsize=13)
plt.ylim(500,610)
plt.tight_layout()
f3=plt.subplot(1,3,3)
plt.bar(df.index,df['第三次月考'],width=0.8)
plt.title('第三次月考成绩对比',color='red',fontsize=13)
plt.ylim(500,610)
plt.tight_layout()
plt.savefig("对比.png",dpi=300,bbox_inches="tight")
plt.show()
```
